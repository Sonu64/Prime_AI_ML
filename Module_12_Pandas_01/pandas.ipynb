#### Series in Pandas --> These are the values in a Single column of a Dataframe ()

```python
# Creating Series from simple Python Lists

print("=========== Index Based Access ===========\n")
mySeries = pd.Series(["Sonu", "Shanu", "Mithu", "Suparna", "Rimi"])
print(mySeries)
# By default, Labels ==> Indices, Accessing via INDICES
print(mySeries[0], mySeries[1])
print("\n")


# LABEL based access, Labels are basically Named INDICES. we may use index=LabelList, but it is actually the list of Labels. INDICES directly point to Positions, LABELS point to the Label associated with that position, Named Indices.
print("=========== Label Based Access ===========\n")
myAgeSeries = pd.Series([22, 23, 45, 19], index=['Sonu', 'Shanu', 'Mithu', 'Suparna'])
print(myAgeSeries)
print(myAgeSeries['Mithu'], myAgeSeries['Shanu']) # Accessing via labels
```

#### Properties of Series

```python
# ➡️ They are Homogenous, Mix of types result in dtype=OBJECT #

# ➡️ They Support vectorized Operations like NumPy Arrays #
s1 = pd.Series([1, 2, 3, 4, 5])
s2 = pd.Series([100, 200, 300, 400, 500])
sum = s1+s2
sum.index = ['Sum of index-0', 'Sum of index-1', 'Sum of index-2', 'Sum of index-3', 'Sum of index-4']
print(sum)

# ➡️ Mutable Values, ➡️ Immutable Size #
series = pd.Series([1, 2, 3, 4, 5])
# Mutable values #
series[2] = 675
print(series)
# Size is immutable, even if we drop something, it returns a new Series with dropped values, original series size does not change #
new_series = series.drop(2) # The Label = 2, entirely Vanishes !! Labels go as 0,1,3,4 🤪🤪
print(series) 
print(new_series)

# ➡️ Handle Missing values with NaN #
s1 = pd.Series([1, 3, 4]) # Index 3, 4 is missing --> NaN
s2 = pd.Series([1, 2, 3, 4, 5]) # All indices filled, extra filled indices are 3 and 4
print(s1+s2) # 3rd and 4th index becomes NaN

```

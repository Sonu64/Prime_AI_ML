#### Series in Pandas → These are the values in a Single column of a Dataframe ()

```python
# Creating Series from simple Python Lists

print("=========== Index Based Access ===========\n")
mySeries = pd.Series(["Sonu", "Shanu", "Mithu", "Suparna", "Rimi"])
print(mySeries)
# By default, Labels ==> Indices, Accessing via INDICES
print(mySeries[0], mySeries[1])
print("\n")


# LABEL based access, Labels are basically Named INDICES. we may use index=LabelList, but it is actually the list of Labels. INDICES directly point to Positions, LABELS point to the Label associated with that position, Named Indices.
print("=========== Label Based Access ===========\n")
myAgeSeries = pd.Series([22, 23, 45, 19], index=['Sonu', 'Shanu', 'Mithu', 'Suparna'])
print(myAgeSeries)
print(myAgeSeries['Mithu'], myAgeSeries['Shanu']) # Accessing via labels
```

#### Properties of Series

```python
# ➡️ They are Homogenous, Mix of types result in dtype=OBJECT #

# ➡️ They Support vectorized Operations like NumPy Arrays #
s1 = pd.Series([1, 2, 3, 4, 5])
s2 = pd.Series([100, 200, 300, 400, 500])
sum = s1+s2
sum.index = ['Sum of index-0', 'Sum of index-1', 'Sum of index-2', 'Sum of index-3', 'Sum of index-4']
print(sum)

# ➡️ Mutable Values, ➡️ Immutable Size #
series = pd.Series([1, 2, 3, 4, 5])
# Mutable values #
series[2] = 675
print(series)
# Size is immutable, even if we drop something, it returns a new Series with dropped values, original series size does not change #
new_series = series.drop(2) # The Label = 2, entirely Vanishes !! Labels go as 0,1,3,4 🤪🤪
print(series) 
print(new_series)

# ➡️ Handle Missing values with NaN #
s1 = pd.Series([1, 3, 4]) # Index 3, 4 is missing --> NaN
s2 = pd.Series([1, 2, 3, 4, 5]) # All indices filled, extra filled indices are 3 and 4
print(s1+s2) # 3rd and 4th index becomes NaN

```

### DataFrames in Pandas → 2D Labelled Arrays, Labels exist for both Rows and Columns

```python
# Creating Dataframes from a List of Pandas Series, from List of dicts already shown firstly !

## Using Uniform labels for every Series ##
rowLabels = ['Index-0', 'Index-1', 'Index-2', 'Index-3']
dataDict1 = {
    'Name': pd.Series(['Sonu', 'Shanu', 'Rimi', 'Suparna'], index=rowLabels),
    'Age': pd.Series([22, 23, 21, 18],index=rowLabels),
    'City': pd.Series(['Chennai', 'Bankura', 'Durgapur', 'Bankura'],index=rowLabels),
    'Favorite Color': pd.Series(['Red', 'Blue', 'Yellow', 'Red'],index=rowLabels)
}

df1 = pd.DataFrame(dataDict1) # Create DataFrame from the dictionary, with labels=rowLabels


## Using Non-Uniform labels for every Series, Pandas UNIONS them all, no error thrown, empty values under a Particular Label gets NaN ## ----> ⚠️ Not Recommended ⚠️
rowLabels1 = ['Zero', 'One', 'Two', 'Three']
rowLabels2 = ['ZERO', 'ONE', 'TWO', 'THREE']

dataDict2 = {
    'Name': pd.Series(['Sonu', 'Shanu', 'Rimi', 'Suparna'], index=rowLabels1),
    'Age': pd.Series([22, 23, 21, 18],index=rowLabels1),
    'City': pd.Series(['Chennai', 'Bankura', 'Durgapur', 'Bankura'],index=rowLabels2),
    'Favorite Color': pd.Series(['Red', 'Blue', 'Yellow', 'Red'],) # GETS DEFAULT INDICES AS LABELS
}

df2 = pd.DataFrame(dataDict2) # Create DataFrame from the dictionary, with labels=rowLabels1, rowLabels2, and one series uses the default indices as Labels
```

```python
df1 # Pandas does not sort the labels, as all Series uses Same Labels
```

```python
df2 # Collects all unique labels from every Series (it makes a "Union"). Automatically sorts them alphabetically/numerically by default during that union process if they aren't identical.
# This sorting is automatic !!
```

```python
# Show the Row Label names --> Indices of the Rows by default
print(df1.index) # RangeIndex(start=0, stop=4, step=1)

# Show the Column Label Names --> The name of the column, the dictionary keys
print(df1.columns) # Index(['Name', 'Age', 'City', 'Favorite Color'], dtype='object')
```

#### Working with Kaggle Datasets, Selecting Data

```python
data = pd.read_csv('globalAirQuality.csv')
df = pd.DataFrame(data)
# df
```

```python
# General Prettify function
def show_heading(string):
    print("="*30, end="")
    print(f" {string} ", end="")
    print("="*30)

# Selecting Columns
# Single Column
show_heading("All Cities")
print(df['city'], end="\n\n")
# Multiple Columns -> Pass List of column labels
show_heading("All Cities and their AQI values")
print(df[['city', 'aqi']], end="\n\n")


# Selecting Rows
# .loc[] syntax --> LOC[row_Labels_or_Slices, column_Labels_or_Slices_IF_PROVIDED] 
# .iloc[] syntax --> ILOC[row_Indices_or_Slices, column_Indices_or_Slices_IF_PROVIDED]
# using loc() takes the Labels (Here labels=indices)
show_heading("Using loc[] to select Rows")
print("\nSelecting Row with Label 4 -------->")
print(df.loc[4])
show_heading("Using loc[start, end, step] to slice df")
print("\nSelection Rows with labels 0 to 4, INCLUSIVE !")
print(df.loc[0:4])
# using iloc() takes the True index values, Here labels=indices, but iloc is EXCLUSIVE !
show_heading("Using iloc[] to select Rows")
print("\nSelecting Row with Index 4 -------->")
print(df.iloc[4])
show_heading("Using iloc[start, end, step] to slice df")
print("\nSelection Rows with labels 0 to 4, EXCLUSIVE !")
print(df.iloc[0:4])


```

```python
# Selecting Rows + Columns ---> A Section of the df, using loc[].
df.loc[0:3, ['city', 'aqi', 'temperature']]
```

```python
# Selecting Rows + Columns ---> A Section of the df, using iloc[].
# iloc can't take String Values, it needs true Integer values to use as indices !
city_index = df.columns.get_loc('city')
aqi_index = df.columns.get_loc('aqi')
temp_index = df.columns.get_loc('temperature')
df.iloc[0:4, [city_index, aqi_index, temp_index]] # TO match above loc[] results we must use end index as 4 here, to include 3, as iloc[] is EXCLUSIVE
```

```python
# Select single scalar value using at() ---> Takes Integer Index values only, for both ROWS and COLS.
df.at[4, 'city']
```

```python
# Select single scalar value using at() ---> Takes true Label values -> Here it is INT for Rows, STRING for cols
df.iat[4, city_index]
```

#### Filtering Data

```python
df[ df['aqi']>100 ] # Show only those rows with aqi > 100
df[ (df['aqi']>100) & (df['country']=="IN")] # Show only Indian cities with aqi > 100

# Chain Filtered rows with columns next --> To see only specified data about the needed cities
desired_data = df[ (df['aqi']>100) & (df['country']=="IN")][['country', 'city', 'aqi', 'temperature']].copy() # ⚠️⚠️⚠️ Don't forget to make a copy ⚠️⚠️⚠️

desired_data.head()
```

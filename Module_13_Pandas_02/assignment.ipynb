```python
##...Working on the Titanic Dataset
df = pd.read_csv('Titanic-Dataset.csv')
df.columns
```

```python
data = df[ (df['Sex']=='female')  & (df['Fare']>30) ].copy()
data
```

```python
total_survived_in_1st_class = len(df[(df['Pclass']==1) & (df['Survived']==1)])
total_survived_in_2nd_class = len(df[(df['Pclass']==2) & (df['Survived']==1)])
total_survived_in_3rd_class = len(df[(df['Pclass']==3) & (df['Survived']==1)])

total_people_in_1st_class = len(df[(df['Pclass']==1)])
total_people_in_2nd_class = len(df[(df['Pclass']==2)])
total_people_in_3rd_class = len(df[(df['Pclass']==3)])

survival_rate_1st_class = (total_survived_in_1st_class / total_people_in_1st_class) * 100
survival_rate_2nd_class = (total_survived_in_2nd_class / total_people_in_2nd_class) * 100
survival_rate_3rd_class = (total_survived_in_3rd_class / total_people_in_3rd_class) * 100


result = df.groupby(['Pclass']).agg(
     Survival_Rate_in_Percentage = ('Survived', 'mean'), # This does the (sum / count) for me !
     Average_Fare = ('Fare', 'mean'),
     Average_Age = ('Age', 'mean')
 )
result['Survival_Rate_in_Percentage'] = result['Survival_Rate_in_Percentage'] * 100

result
```

```python
print(survival_rate_1st_class, survival_rate_2nd_class, survival_rate_3rd_class) # Values match !!
```
